```python
import numpy as np
import pandas as pd
from numpy.linalg import inv, slogdet
from math import log, pi
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
import matplotlib.pyplot as plt

# =======================
# 1. Load and Prepare Data
# =======================
# Assuming dataset is downloaded and extracted in Colab environment
df = pd.read_csv("SMSSpamCollection", sep="\t", header=None, names=["label", "text"])

print("Dataset shape:", df.shape)
print(df.head())

# Encode target labels ('ham' -> 0, 'spam' -> 1)
le = LabelEncoder()
y = le.fit_transform(df['label'])

# Vectorize text using TF-IDF (limit features to 300 to keep covariance manageable)
tfidf = TfidfVectorizer(stop_words='english', max_features=300)
X = tfidf.fit_transform(df['text']).toarray()

print(f"Feature matrix shape after TF-IDF vectorization: {X.shape}")

# Split into train and test sets (stratify to preserve label distribution)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# =======================
# 2. Fit Bayesian Model
# =======================
def fit_gaussian_bayes(X, y, regularize=1e-6):
    """Estimate mean, covariance, and prior for each class."""
    classes = np.unique(y)
    params = {}
    for c in classes:
        Xc = X[y == c]
        mean = Xc.mean(axis=0)
        cov = np.cov(Xc, rowvar=False) + np.eye(Xc.shape[1]) * regularize
        prior = len(Xc) / len(X)
        params[c] = {"mean": mean, "cov": cov, "prior": prior}
    return params

def log_multivariate_normal_pdf(x, mean, cov):
    """Compute log PDF of multivariate Gaussian."""
    d = len(mean)
    x = np.atleast_2d(x)
    xc = x - mean
    sign, logdet = slogdet(cov)
    inv_cov = inv(cov)
    quad = np.sum((xc @ inv_cov) * xc, axis=1)
    return -0.5 * (d * log(2 * pi) + logdet + quad)

def predict_gaussian_bayes(X, params):
    """Predict classes using Bayesian decision rule (MAP)."""
    classes = list(params.keys())
    log_probs = []
    for c in classes:
        p = params[c]
        log_likelihood = log_multivariate_normal_pdf(X, p["mean"], p["cov"])
        log_prior = np.log(p["prior"])
        log_probs.append(log_likelihood + log_prior)
    log_probs = np.column_stack(log_probs)
    return np.array([classes[i] for i in np.argmax(log_probs, axis=1)])

# Train model
params = fit_gaussian_bayes(X_train, y_train)

# Predict
y_pred = predict_gaussian_bayes(X_test, params)

# =======================
# 3. Evaluate Performance
# =======================
accuracy = np.mean(y_pred == y_test)
print("\nAccuracy:", accuracy)

def confusion_matrix(y_true, y_pred):
    classes = np.unique(np.concatenate([y_true, y_pred]))
    matrix = np.zeros((len(classes), len(classes)), dtype=int)
    for t, p in zip(y_true, y_pred):
        i = np.where(classes == t)[0][0]
        j = np.where(classes == p)[0][0]
        matrix[i, j] += 1
    return classes, matrix

classes_, cm = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix (rows=True, cols=Pred):\n", cm)

plt.figure(figsize=(5, 4))
plt.imshow(cm, cmap="Blues", interpolation="nearest")
plt.title("Confusion Matrix - Bayesian Decision Theory")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.xticks(ticks=np.arange(len(classes_)), labels=le.inverse_transform(classes_))
plt.yticks(ticks=np.arange(len(classes_)), labels=le.inverse_transform(classes_))
for (i, j), val in np.ndenumerate(cm):
    plt.text(j, i, val, ha="center", va="center")
plt.tight_layout()
plt.show()

# =======================
# 4. Optional: Save Predictions
# =======================
output = pd.DataFrame(X_test, columns=tfidf.get_feature_names_out())
output["True"] = le.inverse_transform(y_test)
output["Predicted"] = le.inverse_transform(y_pred)
output.to_csv("bayesian_spam_predictions.csv", index=False)
print("\nPredictions saved to bayesian_spam_predictions.csv")
```
